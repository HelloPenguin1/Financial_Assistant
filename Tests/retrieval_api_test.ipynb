```python
from dotenv import load_dotenv
load_dotenv()
import os
```

```python
sec_api_key = os.environ.get("SEC_EDGAR_API_KEY")
```

```python
from sec_api import ExtractorApi  
extractor = ExtractorApi(sec_api_key)  
```

```python
url_10k = "https://www.sec.gov/Archives/edgar/data/1318605/000156459021004599/tsla-10k_20201231.htm&"
risk_factors = extractor.get_section(filing_url = url_10k, section = "1A",return_type = "text")
```

```python
print(risk_factors[:2000])
```

```python
item_ids_10K = [
    "7"
]

for sec in item_ids_10K:
    print(f"Extracting section {sec}...")
    section_text = extractor.get_section(filing_url = url_10k, section = sec, return_type = "html")
    print(f"Section {sec} text: {section_text[:2000]}")
```
